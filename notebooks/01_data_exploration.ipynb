# 01. Full Data Exploration & Profiling: Conflict vs. Economy (AES)
1. Objective
This notebook is a comprehensive End-to-End Exploration of the Mali, Burkina Faso, and Niger datasets. It tracks the transformation of raw data from the ACLED and World Bank sources into actionable insights.

The Workflow:

Data Auditing & Cleaning: Identifying structural issues, handling missing values, and synchronizing timeframes (Weekly ACLED vs. Annual World Bank).

Conflict Profiling: Visualizing the lethality and nature of violence across the AES.

Economic Deep-Dive: Processing GDP, inflation, and agricultural indicators to map regional performance.

Integrative Analysis: Overlapping both datasets to observe the "Conflict-Economy" nexus.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## ACLED DATA

### ACLED data exploring

```python
df = pd.read_csv("../data/raw/Africa_aggregated_data_from_ACLED.csv",sep=";", on_bad_lines='skip')
```

```python
df.head()
```

```python
aes_country = ['Mali', 'Burkina Faso', 'Niger']
df_aes = df[df['COUNTRY'].isin(aes_country)].copy()
```

```python
print(df_aes['EVENT_TYPE'].value_counts())
```

```python
pd.crosstab(df_aes['COUNTRY'], df_aes['EVENT_TYPE'])
```

```python
df_aes['WEEK'] = pd.to_datetime(df_aes['WEEK'], errors='coerce')
df_aes['YEAR'] = df_aes['WEEK'].dt.year
```

```python
df_aes
```

```python
df_aes.drop('year',axis = 1,inplace= True)
```

```python
df_aes = df_aes.reset_index(drop=True)
```

```python
df_aes.describe()
```

```python
print(df_aes['COUNTRY'].value_counts())
```

```python
print(df_aes.groupby('COUNTRY')['FATALITIES'].sum())
```

```python
df_aes = df_aes.drop(columns=['POPULATION_EXPOSURE','REGION'])
```

```python
df_aes
```

### ACLED data vizualisation

```python
# Lineplot for fatalities over time
sns.set_theme(style="ticks")
plt.figure(figsize=(12, 6))

sns.lineplot(data=df_aes, x='YEAR', y='FATALITIES', hue='COUNTRY', 
             estimator='sum', errorbar=None, marker='o', linewidth=2.5)

plt.title('Conflict Intensity: Annual Fatalities in AES (2020-2024)', fontsize=15, pad=20)
plt.ylabel('Total Number of Fatalities')
plt.xlabel('Year')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Country', frameon=True)

plt.show()
```

```python
plt.figure(figsize=(12, 7))

# Aggregate fatalities by event type and country
event_impact = df_aes.groupby(['EVENT_TYPE', 'COUNTRY'])['FATALITIES'].sum().reset_index()

sns.barplot(data=event_impact, x='FATALITIES', y='EVENT_TYPE', hue='COUNTRY')

plt.title('Distribution of Fatalities by Event Type', fontsize=15)
plt.xlabel('Total Fatalities')
plt.ylabel('Type of Violence')
plt.show()
```
